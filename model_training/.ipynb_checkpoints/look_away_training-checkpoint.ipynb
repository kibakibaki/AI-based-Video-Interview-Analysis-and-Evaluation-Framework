# Look Away Model Training

This notebook trains a three-level look-away model without merging ordinary and frequent looking away.

Recommended training approach:

- Predict `look_away_level` directly: 0 = no looking away, 1 = looking away, and 2 = frequently looking away. U is excluded.
- You do not need a separate script for every parameter. A better workflow is to train a baseline in one notebook, then gradually add features, compare models, and tune hyperparameters.
- After the single-label models become stable, consider a multi-label model that predicts multiple behavior labels at the same time.

This notebook loads the window-level training dataset, selects look-away-related features, splits train/test data by video, trains baseline models, evaluates the results, and saves the selected model.

````python
from pathlib import Path

import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    classification_report,
    balanced_accuracy_score,
    f1_score,
    make_scorer,
    precision_score,
    recall_score,
)
from sklearn.model_selection import StratifiedGroupKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists() and (PROJECT_ROOT.parent / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "training" / "window_training_dataset.csv"
MODEL_DIR = PROJECT_ROOT / "models"
MODEL_PATH = MODEL_DIR / "look_away_level_model.joblib"

TARGET = "look_away_level"
CLASS_LABELS = [0, 1, 2]
CLASS_NAMES = ["no_looking_away", "looking_away", "frequent_looking_away"]
````

## 1. Load Dataset

If this file does not exist, run this from the project root first:

```bash
./run_build_training_dataset.sh
```

````python
df = pd.read_csv(DATA_PATH)

print(f"Dataset shape: {df.shape}")
display(df.head())
display(df[TARGET].value_counts(dropna=False).rename("count"))
````

## 2. Choose Look Away Features

This section starts with features directly related to gaze, eye contact, and head movement. More features can be added later for comparison.

````python
LABEL_COLUMNS = [
    "look_away_level",
    "face_not_visible",
    "looking_away",
    "frequent_looking_away",
    "excessive_head_movement",
    "noticeable_fidgeting",
    "distracted_by_offscreen_activity",
    "camera_or_video_issue",
    "reading_from_screen_or_notes",
    "looking_down_frequently",
    "body_out_of_frame",
    "poor_lighting",
    "background_distraction_visible",
    "hand_to_face_or_hair",
]

METADATA_COLUMNS = [
    "filename",
    "window_start",
    "window_end",
    "annotation_quality",
    "notes",
]

LOOK_AWAY_FEATURES = [
    "window_duration",
    "window_frames",
    "eye_contact_ratio",
    "face_visibility_ratio",
    "gaze_center_ratio",
    "valid_gaze_frames",
    "gaze_center_frames",
    "looking_away_total_time",
    "gaze_horizontal_ratio_mean",
    "gaze_horizontal_ratio_std",
    "gaze_vertical_ratio_mean",
    "gaze_vertical_ratio_std",
    "primary_gaze_horizontal_ratio",
    "primary_gaze_vertical_ratio",
    "primary_gaze_zone_ratio",
    "secondary_gaze_zone_ratio",
    "gaze_deviation_from_primary_ratio",
    "gaze_deviation_from_primary_mean",
    "head_pitch_offset_mean",
    "head_pitch_offset_std",
    "head_yaw_mean",
    "head_yaw_std",
    "head_roll_mean",
    "head_roll_std",
    "head_deviation_from_primary_ratio",
    "head_deviation_from_primary_mean",
]

feature_cols = [column for column in LOOK_AWAY_FEATURES if column in df.columns]
missing_features = sorted(set(LOOK_AWAY_FEATURES) - set(feature_cols))

print(f"Using {len(feature_cols)} features")
print(feature_cols)
if missing_features:
    print("Missing features:", missing_features)
````

## 3. Prepare Train/Test Split

This section uses `filename` for group splitting so windows from the same video do not appear in both the training and test sets. This gives a more realistic estimate of generalization.

````python
model_df = df.copy()
model_df[TARGET] = pd.to_numeric(model_df[TARGET], errors="coerce")
model_df = model_df[model_df[TARGET].isin(CLASS_LABELS)].copy()
model_df[TARGET] = model_df[TARGET].astype(int)

X = model_df[feature_cols]
y = model_df[TARGET]
groups = model_df["filename"]

splitter = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
train_idx, test_idx = next(splitter.split(X, y, groups=groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
groups_train = groups.iloc[train_idx]

print(f"Train rows: {len(X_train)}")
print(f"Test rows: {len(X_test)}")
print("Train class distribution:")
display(y_train.value_counts().sort_index().rename("count"))
print("Test class distribution:")
display(y_test.value_counts().sort_index().rename("count"))
````

## 4. Build Baseline Models

`LogisticRegression` is easier to interpret and works well as the first baseline. `RandomForestClassifier` can capture nonlinear patterns and is useful as a second comparison model.

````python
numeric_preprocess = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

logistic_model = Pipeline(
    steps=[
        ("preprocess", numeric_preprocess),
        (
            "classifier",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=RANDOM_STATE,
            ),
        ),
    ]
)

forest_model = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                min_samples_leaf=3,
                class_weight="balanced",
                random_state=RANDOM_STATE,
            ),
        ),
    ]
)

models = {
    "logistic_regression": logistic_model,
    "random_forest": forest_model,
}
````

## 5. Cross Validation On Training Videos

The dataset is still relatively small and level 2 is rare. Macro metrics give each of the three levels equal importance instead of letting level 0 dominate the result.

````python
unique_train_groups = groups_train.nunique()
class_group_counts = model_df.iloc[train_idx].groupby(TARGET)["filename"].nunique()
n_splits = min(5, unique_train_groups, int(class_group_counts.min()))
cv_results = pd.DataFrame()

if n_splits < 2:
    print("Not enough videos for group cross validation. Skipping CV.")
else:
    cv = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=RANDOM_STATE)
    scoring = {
        "precision_macro": make_scorer(precision_score, average="macro", zero_division=0),
        "recall_macro": make_scorer(recall_score, average="macro", zero_division=0),
        "f1_macro": make_scorer(f1_score, average="macro", zero_division=0),
        "balanced_accuracy": "balanced_accuracy",
    }
    cv_rows = []

    for name, model in models.items():
        scores = cross_validate(
            model,
            X_train,
            y_train,
            groups=groups_train,
            cv=cv,
            scoring=scoring,
            error_score="raise",
        )
        cv_rows.append(
            {
                "model": name,
                **{metric: scores[f"test_{metric}"].mean() for metric in scoring},
            }
        )

    cv_results = pd.DataFrame(cv_rows).sort_values("f1_macro", ascending=False)
    display(cv_results)
````

## 6. Fit And Evaluate Selected Model

The model with the best mean macro F1 in grouped cross-validation is selected automatically. The held-out test videos are not used for model selection.

````python
selected_model_name = cv_results.iloc[0]["model"] if not cv_results.empty else "random_forest"
selected_model = models[selected_model_name]

selected_model.fit(X_train, y_train)
y_pred = selected_model.predict(X_test)
y_proba = selected_model.predict_proba(X_test)

print(f"Selected model: {selected_model_name}")
print(classification_report(
    y_test,
    y_pred,
    labels=CLASS_LABELS,
    target_names=CLASS_NAMES,
    zero_division=0,
))
print(f"Balanced accuracy: {balanced_accuracy_score(y_test, y_pred):.3f}")

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=CLASS_LABELS,
    display_labels=["level 0", "level 1", "level 2"],
);
````

## 7. Inspect Per-Level Confidence

A three-class model returns one probability for each level. This section inspects those probabilities without tuning anything on the test set.

````python
probability_df = pd.DataFrame(
    y_proba,
    columns=[f"probability_level_{level}" for level in selected_model.classes_],
    index=X_test.index,
)
probability_df["actual_level"] = y_test
probability_df["predicted_level"] = y_pred
probability_df["filename"] = model_df.loc[X_test.index, "filename"]
display(probability_df.head(20))
````

## 8. Inspect Important Features

This step helps check whether the model learned reasonable signals, such as `eye_contact_ratio`, `gaze_center_ratio`, and `looking_away_total_time`.

````python
classifier = selected_model.named_steps["classifier"]

if hasattr(classifier, "feature_importances_"):
    importance_df = pd.DataFrame(
        {
            "feature": feature_cols,
            "importance": classifier.feature_importances_,
        }
    ).sort_values("importance", ascending=False)
    display(importance_df.head(20))
elif hasattr(classifier, "coef_"):
    coef_df = pd.DataFrame(
        classifier.coef_.T,
        index=feature_cols,
        columns=[f"level_{level}_coefficient" for level in classifier.classes_],
    )
    coef_df["max_abs_coefficient"] = coef_df.abs().max(axis=1)
    display(coef_df.sort_values("max_abs_coefficient", ascending=False).head(20))
````

## 9. Save Model

The saved artifact includes the three-level model, class meanings, feature list, and held-out test video names.

````python
MODEL_DIR.mkdir(parents=True, exist_ok=True)

artifact = {
    "model": selected_model,
    "model_name": selected_model_name,
    "target": TARGET,
    "feature_cols": feature_cols,
    "class_labels": CLASS_LABELS,
    "class_names": CLASS_NAMES,
    "test_videos": sorted(groups.iloc[test_idx].unique().tolist()),
    "data_path": str(DATA_PATH),
}

joblib.dump(artifact, MODEL_PATH)
print(f"Saved model artifact to: {MODEL_PATH}")
````
